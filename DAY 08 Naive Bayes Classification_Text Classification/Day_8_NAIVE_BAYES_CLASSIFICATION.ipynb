step 1: import libraries


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_20newsgroups
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

print("All libraries imported successfully!")

````

step 2: understand dataset

````python
categories=["sci.space","rec.sport.hockey"]
print("selected categories:")

for category in categories:
  print(category)
````

step 3: load the dataset

````python
categories=["sci.space","rec.sport.hockey"]

train_data=fetch_20newsgroups(subset="train",categories=categories,remove=("headers","footers","quotes"),shuffle=True,random_state=42)
test_data=fetch_20newsgroups(subset="test",categories=categories,remove=("headers","footers","quotes"),shuffle=True,random_state=42)

print("training documents:",len(train_data.data))
print("testing documents:",len(test_data.data))
````

````python
print("categories:",train_data.target_names)
````

step 5: mionimal dataset check eda

````python
print("Training data size:", len(train_data.data))
print("Testing data size:", len(test_data.data))

print("\nTraining class distribution:")

class_counts = pd.Series(
    train_data.target
).map(
    dict(enumerate(train_data.target_names))
).value_counts()

print(class_counts)

empty_documents = sum(
    not document.strip()
    for document in train_data.data
)

print("\nEmpty training documents:", empty_documents)
````

step 5: input anfd target features

````python
X_train = train_data.data
y_train = train_data.target

X_test = test_data.data
y_test = test_data.target

target_names = train_data.target_names

print("First document:")
print(X_train[0][:500])

print("\nActual category:", target_names[y_train[0]])
print("Number of classes:", len(target_names))
````

step 7: probability and condition probability

````python
total_documents = 100
space_documents = 60

prior_probability_space = (
    space_documents / total_documents
)

print(
    "Illustrative prior probability of Space:",
    prior_probability_space
)
````

step 8: BAYES theorem

````python
prior_space = 0.5
likelihood_word_given_space = 0.8
likelihood_word = 0.6

posterior_space = (
    likelihood_word_given_space * prior_space
) / likelihood_word

print(
    "Illustrative posterior probability:",
    posterior_space
)
````

step 9: independence assumption

````python
example_features = [
    "space",
    "planet",
    "orbit",
    "satellite"
]

print("Example document features:")

for feature in example_features:
    print(feature)

print(
    "\nNaive Bayes simplifies probability calculations "
    "using the conditional independence assumption."
)
````

STEP 10 — TF-IDF TEXT VECTORIZATION

````python
vectorizer = TfidfVectorizer(
    max_features=10000,
    stop_words="english"
)

X_train_tfidf = vectorizer.fit_transform(X_train)

X_test_tfidf = vectorizer.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)

print(
    "Vocabulary size:",
    len(vectorizer.vocabulary_)
)
````

step11:  multinominal naive bayes

````python
nb_model=MultinomialNB(alpha=1.0)
print(nb_model)
nb_model.fit(X_train_tfidf,y_train)
````

step 12: build ml pipline

````python
model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            max_features=10000,
            stop_words="english"
        )
    ),
    (
        "naive_bayes",
        MultinomialNB(alpha=1.0)
    )
])

print(model)
````

STEP 13 — TRAIN THE MODEL

````python
model.fit(X_train, y_train)

print("Naive Bayes model training completed!")
````

STEP 14 — MAKE PREDICTIONS

````python
y_pred = model.predict(X_test)

predictions = pd.DataFrame({
    "Actual": [
        target_names[label]
        for label in y_test
    ],
    "Predicted": [
        target_names[label]
        for label in y_pred
    ],
    "Document": X_test
})

predictions.head(10)
````

STEP 15 — EVALUATE MODEL PERFORMANCE

````python
print(
    f"Accuracy: {accuracy_score(y_test, y_pred):.4f}"
)

print(
    "Macro Precision:",
    round(
        precision_score(
            y_test, y_pred, average="macro"
        ),
        4
    )
)

print(
    "Macro Recall:",
    round(
        recall_score(
            y_test, y_pred, average="macro"
        ),
        4
    )
)

print(
    "Macro F1-score:",
    round(
        f1_score(
            y_test, y_pred, average="macro"
        ),
        4
    )
)

print("\nClassification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=target_names
    )
)
````

STEP 16 — CONFUSION MATRIX

````python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=target_names,
    yticklabels=target_names
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")
plt.title("Day 8 — Naive Bayes Confusion Matrix")

plt.tight_layout()
plt.show()
````

STEP 17 — ERROR ANALYSIS

````python
errors = predictions[
    predictions["Actual"] != predictions["Predicted"]
]

print("Total test documents:", len(predictions))
print("Incorrect predictions:", len(errors))

for _, row in errors.head(5).iterrows():
    print("\nActual:", row["Actual"])
    print("Predicted:", row["Predicted"])
    print("Document:", row["Document"][:500])
    print("-" * 60)
````

STEP 18 — TEST CUSTOM DOCUMENTS

````python
custom_documents = [
    "NASA launched a spacecraft to explore Mars and distant planets.",
    "The hockey team scored a goal during the final period.",
    "Scientists are studying satellites and space exploration.",
    "The goalie stopped the puck during the hockey match."
]

custom_predictions = model.predict(custom_documents)

custom_probabilities = model.predict_proba(
    custom_documents
)

for document, prediction, probabilities in zip(
    custom_documents,
    custom_predictions,
    custom_probabilities
):
    print("Document:", document)

    print(
        "Predicted category:",
        target_names[prediction]
    )

    for category, probability in zip(
        target_names,
        probabilities
    ):
        print(f"{category}: {probability:.4f}")

    print("-" * 60)
````

STEP 19 — EXPERIMENT WITH SMOOTHING

````python
for alpha in [0.1, 0.5, 1.0, 2.0, 10.0]:

    experiment = Pipeline([
        (
            "tfidf",
            TfidfVectorizer(
                max_features=10000,
                stop_words="english"
            )
        ),
        (
            "naive_bayes",
            MultinomialNB(alpha=alpha)
        )
    ])

    experiment.fit(X_train, y_train)

    experiment_pred = experiment.predict(X_test)

    accuracy = accuracy_score(
        y_test, experiment_pred
    )

    macro_f1 = f1_score(
        y_test,
        experiment_pred,
        average="macro"
    )

    print(
        f"Alpha: {alpha:<4} | "
        f"Accuracy: {accuracy:.4f} | "
        f"Macro F1: {macro_f1:.4f}"
    )
````

STEP 20 — COMPARE NAIVE BAYES VARIANTS

````python
from sklearn.naive_bayes import (
    GaussianNB,
    MultinomialNB,
    BernoulliNB
)

print("GaussianNB:", GaussianNB())
print("MultinomialNB:", MultinomialNB())
print("BernoulliNB:", BernoulliNB())

print(
    "\nSelected model for Day 8: MultinomialNB"
)
````
