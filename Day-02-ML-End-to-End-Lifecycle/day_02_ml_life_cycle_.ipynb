```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.scatter(df["Area"],df["Price"])
plt.xlabel("area")
plt.ylabel("price")
plt.title("area vs price")
plt.show()
```

```python
sns.boxplot(x="Location",y="Price",data=df)
plt.show()
```

interpretation step:
Price distribution is fairly spread out — values range roughly from ₹50K to ₹1M, with no single dominant price range.

Location doesn't show a huge difference in price — Downtown, Suburban, Urban, and Rural have fairly similar price ranges.

Suburban has a slightly higher median price compared with the other locations.

Area vs Price shows no clear pattern — larger area doesn't consistently mean higher price in this dataset.

The scatter plot is widely spread, which suggests that Area alone may not be enough to predict house price.

The target distribution has no obvious extreme outliers, but we should still check the data statistically before making a final

conclusion.

observation:
From these plots, we can see that price varies quite a lot. Location seems to have some variation, but the differences are not very strong. More importantly, Area and Price don't show a clear relationship here, so we cannot depend on Area alone. This tells us that other features like bedrooms, bathrooms, year built, condition, and garage may also play an important role.”
